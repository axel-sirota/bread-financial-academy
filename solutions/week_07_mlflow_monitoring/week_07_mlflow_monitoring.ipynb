## Key Takeaways

### 1. Deployment != Training
Training produces a model artifact. Deployment makes it **useful** by hosting it on infrastructure that accepts requests.

### 2. Monitor Everything
Production ML requires two types of monitoring:
- **Operational**: Is the endpoint running? Are requests failing? Is latency acceptable?
- **Model**: Is the data drifting? Is model performance degrading?

### 3. Experiment Tracking is Non-Negotiable
Without MLflow (or similar), you'll forget:
- Which model version is in production
- What hyperparameters you used
- When you deployed and why

### 4. Drift Happens
Models don't break loudly - they degrade silently. Model Monitor catches this before customers notice.

### 5. Cost Control Matters
A forgotten endpoint can cost hundreds of dollars per month. Always:
- Set CloudWatch alarms for unexpected usage
- Delete endpoints after testing
- Use auto-scaling for production workloads

### 6. Real-Time vs Batch
Real-time endpoints are **always on**, waiting for requests:
- Best for: User-facing applications, sub-second latency requirements
- Cost: Pay for uptime, even with no traffic
- Alternative: **Batch Transform** for large-scale predictions on schedule

# Part 1: Model Deployment to Real-Time Endpoint

## What is a SageMaker Endpoint?

A **SageMaker Endpoint** is a fully-managed service that hosts your trained model for real-time inference. Think of it as a REST API that:
- Accepts prediction requests (features in → predictions out)
- Scales automatically based on traffic
- Handles infrastructure (servers, load balancing, health checks)
- Costs money only while running (~$0.05-0.30/hour depending on instance type)

## Deployment Architecture

```
Your Application
       ↓ HTTP Request
   [API Gateway] (optional)
       ↓
 SageMaker Endpoint
       ↓
   Model Container (XGBoost)
       ↓ Prediction
Your Application ← Response
```

## Key Deployment Concepts

| Component | Purpose | Week 7 Configuration |
|-----------|---------|----------------------|
| **Model** | The trained artifact (.tar.gz) | Week 6 XGBoost fraud model |
| **Instance Type** | Compute resources | ml.t2.medium (2 vCPU, 4GB RAM) |
| **Instance Count** | Number of servers | 1 (for demo, production uses 2+) |
| **Serializer** | Format input data | CSVSerializer (features as CSV) |
| **Deserializer** | Parse response | JSONDeserializer (predictions as JSON) |
| **Data Capture** | Record requests/responses | Enabled (for Model Monitor) |

## Instance Types & Costs

Choosing the right instance type balances **cost** vs **latency**:

| Instance Type | vCPU | Memory | Cost/Hour | Best For |
|---------------|------|--------|-----------|----------|
| ml.t2.medium | 2 | 4 GB | ~$0.05 | Development, low-traffic |
| ml.m5.large | 2 | 8 GB | ~$0.12 | Production, moderate traffic |
| ml.c5.xlarge | 4 | 8 GB | ~$0.20 | High-throughput, CPU-intensive |
| ml.p3.2xlarge | 8 | 61 GB | ~$3.80 | Deep learning models (GPU) |

💡 **Cost Control Tip**: Always delete endpoints after testing! A forgotten endpoint running for a month can cost $36-720.

# Part 2: Real-Time Inference Testing

## How Inference Works

Once deployed, your endpoint accepts **HTTP POST requests** containing feature data and returns predictions.

### Request/Response Flow

```
1. Client prepares features → [0.15, 0.72, 5, 8, 0, 1, 0, ...]
2. Serializer formats data → CSV string: "0.15,0.72,5,8,0,1,0,..."
3. Send HTTP POST → SageMaker Endpoint
4. Model processes features → XGBoost inference
5. Return prediction → {"predictions": [0.85]}
6. Deserializer parses JSON → 0.85 (fraud probability)
```

### Prediction Interpretation

For our fraud detection model:
- **Output**: Probability between 0 and 1
- **< 0.5**: Predicted as NOT fraud (class 0)
- **>= 0.5**: Predicted as fraud (class 1)
- **Threshold**: Can be tuned based on business requirements (false positive vs false negative trade-off)

### Inference Metrics to Track

| Metric | Description | Good Target |
|--------|-------------|-------------|
| **Latency** | Time from request to response | < 100ms (p99) |
| **Throughput** | Requests per second | Depends on traffic |
| **Error Rate** | % of failed requests | < 0.1% |
| **Model Latency** | Just the inference step | < 50ms |

## Lab 2: Batch Inference and Performance Analysis

Process all test samples and calculate performance metrics.

### Your Task

1. Loop through all samples in `test_df`
2. Make a prediction for each sample
3. Record: prediction, probability, latency
4. Calculate summary statistics:
   - Average latency
   - P50, P95, P99 latency (percentiles)
   - Fraud rate (% predicted as fraud)
5. Create a DataFrame with all results

### Expected Output

- Predictions for all test samples
- Latency distribution
- Summary statistics table
- Fraud prediction distribution

### Hints

- Use a list to collect results, then convert to DataFrame
- Latency: `start = time.time()`, then `(time.time() - start) * 1000` for milliseconds
- Percentiles: `np.percentile(latencies, [50, 95, 99])`

````python
# =============================================================================
# LAB 2 SOLUTION: BATCH INFERENCE AND PERFORMANCE ANALYSIS
# =============================================================================

print("Running batch inference on all test samples...")
print("=" * 60)

results = []

if test_df is not None:
    # SOLUTION: Loop through all rows in the test DataFrame
    # iterrows() gives us index and row as a Series
    for idx, row in test_df.iterrows():
        
        # SOLUTION: Convert row to list of feature values
        # This is what the model expects as input
        features = row.values.tolist()
        
        # SOLUTION: Make prediction and measure latency
        # Capture start time before prediction
        start_time = time.time()
        
        # Call the predictor with features
        # The predictor handles CSV serialization and JSON deserialization
        prediction = predictor.predict(features)
        
        # Calculate latency in milliseconds
        latency_ms = (time.time() - start_time) * 1000
        
        # SOLUTION: Extract fraud probability from prediction response
        # XGBoost returns predictions in this format: {'predictions': [{'score': 0.85}]}
        fraud_prob = prediction['predictions'][0]['score']
        
        # SOLUTION: Apply threshold to get binary prediction
        # 0.5 is our decision boundary (can be tuned based on business needs)
        predicted_class = 1 if fraud_prob >= 0.5 else 0
        
        # Store results in dictionary
        results.append({
            'sample_id': idx,
            'fraud_probability': fraud_prob,
            'predicted_class': predicted_class,
            'latency_ms': latency_ms
        })
        
        # Progress indicator every 5 samples
        if (idx + 1) % 5 == 0:
            print(f"  Processed {idx + 1}/{len(test_df)} samples...")
    
    # SOLUTION: Convert results list to DataFrame for analysis
    results_df = pd.DataFrame(results)
    
    print(f"\n✅ Batch inference complete: {len(results_df)} predictions")
    
    # SOLUTION: Calculate summary statistics
    # Average latency across all predictions
    avg_latency = results_df['latency_ms'].mean()
    
    # SOLUTION: Calculate percentiles using numpy
    # P50 (median): 50% of requests are faster than this
    p50_latency = np.percentile(results_df['latency_ms'], 50)
    
    # P95: 95% of requests are faster than this (important SLA metric)
    p95_latency = np.percentile(results_df['latency_ms'], 95)
    
    # P99: 99% of requests are faster than this (tail latency)
    p99_latency = np.percentile(results_df['latency_ms'], 99)
    
    # SOLUTION: Calculate fraud rate (percentage predicted as fraud)
    fraud_rate = results_df['predicted_class'].sum() / len(results_df) * 100
    
    print("\n" + "=" * 60)
    print("Performance Summary:")
    print("=" * 60)
    print(f"Total Predictions:    {len(results_df)}")
    print(f"Average Latency:      {avg_latency:.2f} ms")
    print(f"P50 Latency:          {p50_latency:.2f} ms")
    print(f"P95 Latency:          {p95_latency:.2f} ms")
    print(f"P99 Latency:          {p99_latency:.2f} ms")
    print(f"\nFraud Predictions:    {results_df['predicted_class'].sum()} ({fraud_rate:.1f}%)")
    print(f"Normal Predictions:   {(results_df['predicted_class'] == 0).sum()} ({100-fraud_rate:.1f}%)")
    
    print("\n" + "=" * 60)
    print("Sample Predictions:")
    print("=" * 60)
    display(results_df.head(10))
    
    # SOLUTION NOTES:
    # - We measured end-to-end latency (includes network + serialization + inference)
    # - In production, you'd also track model latency separately (just inference)
    # - P95/P99 are crucial for SLAs - they show worst-case performance
    # - Fraud rate helps validate model behavior (should match training distribution)
    
else:
    print("⚠️ No test data available")

print("=" * 60)
````

# Part 4: Model Monitor - Automated Drift Detection

## Why Monitor Models?

ML models degrade over time due to:

### 1. Data Drift
**Input distributions change**, but the relationship between features and target stays the same.

Example: Transaction amounts increase from $50 avg to $150 avg due to inflation.
- Model still works, but sees "unusual" data
- May reduce confidence or accuracy

### 2. Concept Drift
**Relationships change** - what predicts fraud evolves.

Example: Fraudsters adapt tactics, new fraud patterns emerge.
- Model becomes outdated
- Accuracy drops significantly

### 3. Model Staleness
Model assumptions no longer hold.

Example: Trained on data from 2023, deployed in 2025.
- Customer behavior changed
- New payment methods introduced

## SageMaker Model Monitor

**Automatic drift detection** without manual intervention:

```
Real Requests → Endpoint → Model Monitor
                             ↓
                    Compare to Baseline
                             ↓
                    Detect Violations
                             ↓
                    Trigger Alerts
```

## How Model Monitor Works

| Step | Description | Output |
|------|-------------|--------|
| **Baseline Creation** | Analyze training data statistics | Mean, std, min, max per feature |
| **Data Capture** | Record endpoint requests/responses | Stored in S3 |
| **Scheduled Monitoring** | Compare live data to baseline | Violations report |
| **Constraint Violations** | Flag when drift detected | CloudWatch metrics, S3 reports |

## Types of Drift Detection

1. **Feature Drift**: Input feature distributions change
2. **Prediction Drift**: Output distribution changes
3. **Data Quality**: Missing values, type mismatches

## Complete Code: Create CloudWatch Alarms

This section provides complete, working code (no lab) to set up operational monitoring.

````python
# =============================================================================
# CREATE CLOUDWATCH ALARMS FOR ENDPOINT MONITORING
# =============================================================================

print("Creating CloudWatch alarms for endpoint monitoring...")
print("=" * 60)

# First, create an SNS topic for notifications (optional but recommended)
try:
    topic_name = f"sagemaker-endpoint-alerts-{datetime.now().strftime('%Y%m%d')}"
    
    # Create SNS topic
    topic_response = sns.create_topic(Name=topic_name)
    topic_arn = topic_response['TopicArn']
    
    print(f"✅ SNS Topic created: {topic_name}")
    print(f"   ARN: {topic_arn}")
    print(f"\n💡 To receive email alerts, subscribe to this topic:")
    print(f"   aws sns subscribe --topic-arn {topic_arn} --protocol email --notification-endpoint your@email.com")
    
except sns.exceptions.TopicLimitExceededException:
    print(f"⚠️ SNS topic limit reached, using console to find existing topic")
    topic_arn = None
except Exception as e:
    print(f"⚠️ Could not create SNS topic: {e}")
    topic_arn = None

# Create CloudWatch alarms
print(f"\n📊 Creating CloudWatch alarms...")

# Alarm 1: High Model Latency
try:
    cloudwatch.put_metric_alarm(
        AlarmName=f"{endpoint_name}-HighLatency",
        ComparisonOperator='GreaterThanThreshold',
        EvaluationPeriods=2,
        MetricName='ModelLatency',
        Namespace='AWS/SageMaker',
        Period=300,  # 5 minutes
        Statistic='Average',
        Threshold=1000.0,  # 1 second
        ActionsEnabled=True if topic_arn else False,
        AlarmActions=[topic_arn] if topic_arn else [],
        AlarmDescription='Alert when model latency exceeds 1 second',
        Dimensions=[
            {'Name': 'EndpointName', 'Value': endpoint_name},
            {'Name': 'VariantName', 'Value': 'AllTraffic'}
        ]
    )
    print(f"  ✅ High Latency alarm: {endpoint_name}-HighLatency")
except Exception as e:
    print(f"  ⚠️ Could not create latency alarm: {e}")

# Alarm 2: 4XX Errors (Client Errors)
try:
    cloudwatch.put_metric_alarm(
        AlarmName=f"{endpoint_name}-4XXErrors",
        ComparisonOperator='GreaterThanThreshold',
        EvaluationPeriods=1,
        MetricName='Invocation4XXErrors',
        Namespace='AWS/SageMaker',
        Period=300,
        Statistic='Sum',
        Threshold=5.0,  # More than 5 errors in 5 minutes
        ActionsEnabled=True if topic_arn else False,
        AlarmActions=[topic_arn] if topic_arn else [],
        AlarmDescription='Alert on client errors (bad requests)',
        Dimensions=[
            {'Name': 'EndpointName', 'Value': endpoint_name},
            {'Name': 'VariantName', 'Value': 'AllTraffic'}
        ]
    )
    print(f"  ✅ 4XX Errors alarm: {endpoint_name}-4XXErrors")
except Exception as e:
    print(f"  ⚠️ Could not create 4XX alarm: {e}")

# Alarm 3: Low Invocations (Endpoint might be down)
try:
    cloudwatch.put_metric_alarm(
        AlarmName=f"{endpoint_name}-LowTraffic",
        ComparisonOperator='LessThanThreshold',
        EvaluationPeriods=2,
        MetricName='Invocations',
        Namespace='AWS/SageMaker',
        Period=600,  # 10 minutes
        Statistic='Sum',
        Threshold=1.0,  # Less than 1 invocation in 10 minutes
        ActionsEnabled=True if topic_arn else False,
        AlarmActions=[topic_arn] if topic_arn else [],
        AlarmDescription='Alert when endpoint receives no traffic',
        Dimensions=[
            {'Name': 'EndpointName', 'Value': endpoint_name},
            {'Name': 'VariantName', 'Value': 'AllTraffic'}
        ],
        TreatMissingData='notBreaching'  # Don't alarm if no data points
    )
    print(f"  ✅ Low Traffic alarm: {endpoint_name}-LowTraffic")
except Exception as e:
    print(f"  ⚠️ Could not create low traffic alarm: {e}")

print(f"\n" + "=" * 60)
print(f"✅ CloudWatch monitoring configured!")
print(f"=" * 60)
print(f"\n📈 View metrics in CloudWatch:")
print(f"   https://console.aws.amazon.com/cloudwatch/home?region={region}#alarmsV2:")
print(f"\n💡 These alarms will:")
print(f"   • Monitor endpoint health 24/7")
print(f"   • Alert on latency spikes, errors, or downtime")
print(f"   • Send notifications to SNS topic (if configured)")
print("=" * 60)
````
